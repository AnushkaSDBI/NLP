```python
import spacy

nlp = spacy.load('en_core_web_sm')

text = 'Researchers analyze large datasets efficiently'

doc = nlp(text)

for token in doc:
    print(
        f"{token.text:<15}"
        f"{token.dep_:<15}"
        f"{token.doc.text}"
    )
```

```python
from spacy import displacy
from IPython.display import HTML, display_svg
html = displacy.render(doc, style='dep', jupyter = False)
display(HTML(html))
```

```python
text3 = 'Apple opened a new office in Berlin in 2024.'
doc3 = nlp(text3)
for ent in doc3.ents:
    print(ent.text, ent.label_)
```

```python
print(f"{'Entity':<20} {'Label':<10} {'Explanation'}")
print("-" * 50)

for ent in doc3.ents:
    print(f"{ent.text:<20} {ent.label_:<10} {spacy.explain(ent.label_)}")
```

```python
# Import Hamming distance function
from scipy.spatial import distance

# Define two strings of equal length
string_1 = "table"
string_2 = "taken"

# Calculate normalized Hamming distance
# Result = differing positions / total positions
hamming_distance = distance.hamming(
    list(string_1),
    list(string_2)
)

print("Normalized Hamming Distance:",hamming_distance)

# Convert normalized distance to actual number of differing characters
actual_distance = hamming_distance * len(string_1)

print("Actual Hamming Distance:", actual_distance)
```

```python
doc5 = 'I love natural language processing'
doc6 = 'I love machine learning'

words5 = doc5.lower().split()
words6 = doc6.lower().split()

set5 = set(words5)
set6 = set(words6)

intersection = set5.intersection(set6)
union = set5.union(set6)

ans = len(intersection)/len(union)
print(ans)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


documents = [
     'I love natural language processing',
     'I love machine learning',
     'Machine learning is powerful'
]

vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(documents)

print('Vocab', vectorizer.get_feature_names_out())

similarity_matrix = cosine_similarity(tfidf_matrix)
print('Similarity Matrix', similarity_matrix)

similarity_score = similarity_matrix(0)[1]
print('Cosine similarity score', round(similarity_score, 3))
```
