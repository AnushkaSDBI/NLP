```python
text = 'The sky is clear and the stars are twinking in the night'
tokens =  word_tokenize(text)
stop_words = set(stopwords.words('english'))
filteredwords = [
    word for word in tokens
    if word.lower() not in stop_words
]
print(filteredwords)
```

```python
from nltk.stem import PorterStemmer
from nltk.stem import SnowballStemmer
from nltk.stem import LancasterStemmer
```

```python
words = ['running', 'studies', 'happily', 'university', 'relational', 'relationally', 'happiness']
porter = PorterStemmer()
snowball = SnowballStemmer('english')
lancaster = LancasterStemmer()
print("porter:", [porter.stem(word) for word in words])
print("snowball:", [snowball.stem(word) for word in words])
print("lancaster:", [lancaster.stem(word) for word in words])
```

```python
import nltk
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
nltk.download('omw-1.4')
lemmatizer = WordNetLemmatizer()
words = ['running', 'studies', 'better', 'wolves', 'children']
print([lemmatizer.lemmatize(word) for word in words])
```

```python
import spacy
from langdetect import detect, LangDetectException

# --- Sample complaints ---
complaints = [
    "OMGGG 😭😭 my flight from FRA → DEL got delayed AGAIN!!! #Lufthansa @airporthelp",
    "The baggage tracking system isn't working properly after the latest update :(",
    "GREAT SERVICE!!! waited ONLY 5 hours for support 👏👏 #sarcasm",
    "Sehr schlechter Kundenservice!!! Niemand antwortet auf meine Emails.",
    "Flight AI-202 cancelled 😡 Refund still pending after 3 weeks!!!",
    "Visit www.airlinehelp.com/refund to check your refund status.",
    "I can't login to the app since update v2.4.1 pls fix ASAP!!!!",
    "😠😡",
    "Seats were fine but food was meh...",
    "SUPPORT SUPPORT SUPPORT!!!!!!!",
]

# --- Load spaCy models once (not inside the loop) ---
print("Loading spaCy models...")
nlp_en = spacy.load("en_core_web_sm")
nlp_de = spacy.load("de_core_news_sm")
print("Models loaded.\n")

def pick_model(text: str):
    """Detect language and return the matching spaCy model + language name."""
    try:
        lang = detect(text)
    except LangDetectException:
        lang = "en"          # fallback for emoji-only / garbage text

    if lang == "de":
        return nlp_de, "German"
    else:
        return nlp_en, "English"   # default everything else to English

def process_complaint(text: str):
    nlp, lang_name = pick_model(text)
    doc = nlp(text)

    tokens   = [t.text    for t in doc if not t.is_space]
    lemmas   = [t.lemma_  for t in doc if not t.is_space and not t.is_punct]
    entities = [(ent.text, ent.label_) for ent in doc.ents]

    return {
        "text":     text,
        "language": lang_name,
        "tokens":   tokens,
        "lemmas":   lemmas,
        "entities": entities,
    }

# --- Run ---
print("=" * 65)
for i, complaint in enumerate(complaints, start=1):
    result = process_complaint(complaint)

    print(f"[{i}] {result['text']}")
    print(f"     Language : {result['language']}")
    print(f"     Tokens   : {result['tokens']}")
    print(f"     Lemmas   : {result['lemmas']}")
    print(f"     Entities : {result['entities'] if result['entities'] else 'none found'}")
    print("-" * 65)
```
